```python
import pandas as pd

df = pd.read_csv("telco_churn.csv")
print(df.shape)
print(df["Churn"].value_counts(normalize=True))
```

```python
from sklearn.model_selection import train_test_split

# Encoding fitur kategorikal
df = pd.get_dummies(df, drop_first=True)

# Pisahkan fitur (X) dan target (y)
X = df.drop("Churn_Yes", axis=1)
y = df["Churn_Yes"]

X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42
)

rf.fit(X_tr, y_tr)
```

```python
from sklearn.metrics import classification_report
from sklearn.metrics import roc_auc_score

# Hitung prediksi
y_pred = rf.predict(X_te)

# Hitung probabilitas kelas churn
y_prob = rf.predict_proba(X_te)[:, 1]

# Tampilkan hasil evaluasi
print(classification_report(y_te, y_pred))

print("ROC-AUC:", roc_auc_score(y_te, y_prob))
```

```python
# Hitung probabilitas churn
prob_churn = rf.predict_proba(X_te)[:, 1]

# Tampilkan 10 probabilitas pertama
print(prob_churn[:10])
```

Model Random Forest menunjukkan performa yang baik dalam memprediksi pelanggan yang berpotensi melakukan churn berdasarkan nilai precision, recall, F1-score, dan ROC-AUC yang diperoleh. Selain menghasilkan prediksi kelas, model juga memberikan probabilitas churn melalui `predict_proba`, sehingga tingkat risiko setiap pelanggan dapat diketahui. Pelanggan dengan probabilitas mendekati 1 memiliki risiko churn yang lebih tinggi dan dapat diprioritaskan untuk program retensi. Dengan demikian, model dapat membantu perusahaan dalam mengambil keputusan yang lebih tepat untuk mengurangi tingkat churn.
