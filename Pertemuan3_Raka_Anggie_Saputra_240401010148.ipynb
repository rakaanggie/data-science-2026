```python
import pandas as pd
import numpy as np
from scipy.stats.mstats import winsorize
```

```python
# Step 1 (3) - Muat Dataset
url = 'https://drive.usercontent.google.com/download?id=1LfQWProB0VjWN5q8bKuRIgn-stULfIRo&export=download&authuser=0'
df = pd.read_csv(url)
```

```python
# Step 2 (4) - Eksplorasi Info, Describe, dan Missing
print('\nInfo dataset: \n',df.info())
print('\nStatistik dataset: \n',df.describe())
print('\nTotal missing: \n',df.isnull().sum())
```

```python
# Step 3 (5) - Hapus Baris Duplikat
df.drop_duplicates(inplace=True)
print('Jumlah duplikat:', df.duplicated().sum())
```

```python
# Step 4 (6) - Normalisasi
df['kota'] = df['kota'].str.strip().str.title()
df['kondisi'] = df['kondisi'].str.strip().str.lower()
```

```python
# Step 5 (7) - Imputasi Missing Values
df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())
df['harga_juta'] = df['harga_juta'].fillna(df['harga_juta'].median())
df['kamar'] = df['kamar'].fillna(df['kamar'].mode() [0])
```

```python
# Step 6 (8) - Penanganan Outliers
for col in ['harga_juta', 'luas_m2', 'tahun_bangun']:
  Q1, Q3 = df[col].quantile([0.25, 0.75])
  IQR = Q3 -Q1
  df[col] = df[col].clip(Q1-1.5*IQR, Q3+1.5*IQR)
```

```python
# Step 7 (9) - Validasi
assert df.isnull().sum().sum() == 0, 'Masih ada missing!'
assert df.duplicated().sum().sum() == 0, 'Masih ada duplikat!'
print("Total missing values:", df.isnull().sum().sum())
print("Total duplikat:", df.duplicated().sum())
print('Shape akhir:', df.shape)
```

```python
df.to_csv('housing_clean.csv', index=False)
print('Dataset bersih telah tersimpan!')
```
